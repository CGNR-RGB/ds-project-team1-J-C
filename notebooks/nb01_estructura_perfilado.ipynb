### **Columnas**
- **folio**: identificador que nos permite reconocer cada incidente que se ha reportado
- **fecha_hora_creacion**: fecha exacta en la que se generó el reporte
- **fecha_hora_cierre**: fecha en la que se concluyó la atención del reporte y se registró en el sistema
- **tipo_incidente_c4**: Se reporta una categoría de accidente e.g. Accidente, Lesionado
- **incidente_c4**: se agregamás detalles del incidente, e.g si hubieron motociclistas 
- **alcaldia_inicio**: alcaldía donde ocurrió el incidente
- **codigo_cierre**: Es una clasificación que divide cada incidente en cuatro letras {A,F,D,I} y se refieren a A = “Afirmativo” F = “Falso” D = “Duplicados” I = “Informativo”
- **clas_con_f_alarma**
- **tipo_entrada**
- **alcaldia_cierre**: 
- **alcaldia_catalogo**: Nombre de la alcaldía, aunque deberían ser 16 algunos registros vienen con na
- **colonia_catalogo**: se incluyende los nombre de XXXXXXX colonias
- **longitud**: coordenadas de la longitud
- **latitud**: coordenadas de la latitud
